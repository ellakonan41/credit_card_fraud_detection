```python
df['Hour'] = (df['Time'] // 3600) % 24
```

```python
df['is_night'] = df['Hour'].between(0, 6).astype(int)
```

## Synthèse — Feature Engineering (étape 2)

### Features créées

| Feature | Origine | Description | Validation |
|---|---|---|---|
| **Hour** | `Time` | Heure de la journée (0-23), extraite via `(Time // 3600) % 24` | Taux de fraude variable selon l'heure, pics à 2h/4h (>1.6% et 1%, ~9-10x le taux moyen) |
| **Amount_log** | `Amount` | `log(Amount + 1)`, pour compresser l'asymétrie forte du montant | Distribution plus lisible/symétrique (confirmé visuellement en EDA) ; garde `Amount` original en parallèle pour l'interprétabilité |
| **is_night** | `Hour` | Binaire (0/1), vaut 1 si Hour entre 0h et 6h | Taux de fraude 3.4x plus élevé la nuit (0.47%) que le jour (0.14%) — résume le pattern temporel en une seule variable directement exploitable |

### Limite identifiée
Pas de feature de fréquence/historique de transaction possible : le dataset est
anonymisé sans identifiant client, donc impossible de calculer par exemple
"nombre de transactions du même client dans la dernière heure" — à assumer
comme limite du dataset, pas un oubli du feature engineering.

### Variables déjà présentes jugées pertinentes (issues de l'EDA, pas de
